**1. Introduction & Problem Statement**

**Problem:** In data science, before building any predictive model, we must understand the underlying structure of the data. The goal of this task is to perform an Initial Data Inspection and Exploratory Data Analysis (EDA) on a classic biological dataset. We aim to identify patterns and relationships between physical measurements of flowers to see if they naturally group into different species.

**2. Dataset Understanding & Description**

The Iris Dataset is a multivariate dataset introduced by British biologist Ronald Fisher. It is the "Hello World" of data science.
**Target Variable**: species (Setosa, Versicolor, and Virginica).
Features (Attributes):
sepal_length & sepal_width: Measurements of the outer protective part of the flower.
petal_length & petal_width: Measurements of the inner colorful part of the flower.
**Structure:** 150 instances (rows) with 4 numerical features and 1 categorical label.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset directly from seaborn's built-in library
df = sns.load_dataset('iris')

print("Libraries loaded and dataset imported successfully!")
```

```python
# 1. Show the shape (Dimensions)
print("Dataset Shape:", df.shape)

# 2. Show the column names
print("\nColumns:", df.columns.tolist())

# 3. Display the first 5 rows
df.head()
```

```python
# 1. Check for missing values (important for data integrity)
print("Missing values in each column:")
print(df.isnull().sum())

# 2. Get a statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species', s=100)
plt.title('Petal Length vs Petal Width')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['sepal_length'], kde=True, color='teal')
plt.title('Distribution of Sepal Length')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='species', y='sepal_width', data=df)
plt.title('Sepal Width Distribution by Species')
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(15, 6))

# Plot 1: Sepal (The "Messy" one)
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species', ax=ax[0])
ax[0].set_title('Sepal Length vs Width (More Overlap)')

# Plot 2: Petal (The "Clear" one)
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species', ax=ax[1])
ax[1].set_title('Petal Length vs Width (Clearer Clusters)')

plt.show()
```
